## Organise Common and Non-Common N-gram Files

Loop through all data types, corpora, models, and raw directories to reorganise `.rds` files into `common` and `non_common` subdirectories.

Remove the corresponding suffix from filenames, skip missing or empty directories, and avoid overwriting existing files.

```python
from pathlib import Path
```

```python
BASE_DIR = Path(
    "/Volumes/BCross/av_datasets_updated_experiments/common_and_uncommon"
)
```

```python
moved = 0
skipped = 0

# Loop through data types
for data_type_dir in BASE_DIR.iterdir():
    if not data_type_dir.is_dir():
        continue

    # Loop through corpora
    for corpus_dir in data_type_dir.iterdir():
        if not corpus_dir.is_dir():
            continue

        # Loop through models
        for model_dir in corpus_dir.iterdir():
            if not model_dir.is_dir():
                continue

            # Loop through raw directories (raw_10, etc.)
            for raw_dir in model_dir.glob("raw_*"):
                if not raw_dir.is_dir():
                    continue

                # Only files directly inside raw_dir
                files = list(raw_dir.glob("*.rds"))

                if not files:
                    continue

                for file in files:

                    if file.stem.endswith("_non_common"):
                        subfolder = "non_common"
                        new_stem = file.stem.removesuffix("_non_common")

                    elif file.stem.endswith("_common"):
                        subfolder = "common"
                        new_stem = file.stem.removesuffix("_common")

                    else:
                        skipped += 1
                        continue

                    destination_dir = raw_dir / subfolder
                    destination_dir.mkdir(parents=True, exist_ok=True)

                    destination = destination_dir / f"{new_stem}.rds"

                    # Avoid overwriting existing files
                    if destination.exists():
                        skipped += 1
                        continue

                    file.rename(destination)
                    moved += 1

print(f"Files moved: {moved}")
print(f"Files skipped: {skipped}")
```
