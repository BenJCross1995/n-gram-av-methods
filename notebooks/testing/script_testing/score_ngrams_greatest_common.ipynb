```python
import os
import subprocess
import sys

from pathlib import Path

from from_root import from_root

sys.path.insert(0, str(from_root("src")))

from read_and_write_docs import read_rds, read_jsonl
from utils import apply_temp_doc_id, build_metadata_df
```

```python
# --- assume notebook is in: n-gram-av-methods/notebooks/script_testing/ ---
REPO_ROOT = Path.cwd().resolve().parents[2]

SCRIPT = REPO_ROOT / "scripts" / "score_ngrams_multi.py"

PYTHON = REPO_ROOT / "my_venv/bin/python"
```

```python

# --- set your args (strings) ---
base_loc = "/Volumes/BCross"

if not os.path.exists(base_loc):
    print("Using Offline Location")
    base_loc = "/Users/user/Documents/uni_work_offline"
else:
    print("Using Volume Location")

CORPUS = "All-the-news"
DATA_TYPE = "test"
PROBLEM = "William Saunderson-Meyer vs William Saunderson-Meyer"

KNOWN_LOC = f"{base_loc}/datasets/author_verification/{DATA_TYPE}/{CORPUS}/known_raw.jsonl"
UNKNOWN_LOC = f"{base_loc}/datasets/author_verification/{DATA_TYPE}/{CORPUS}/unknown_raw.jsonl"
METADATA_LOC = f"{base_loc}/datasets/author_verification/{DATA_TYPE}/metadata.rds"
MODEL_LOC = f"{base_loc}/models/gpt2"

SAVE_LOC = f"{base_loc}/av_datasets_experiments/concat_greatest_common_ngram/new"
COMPLETED_LOC = f"{base_loc}/av_datasets_experiments/concat_greatest_common_ngram/new"
SCORE_SCRIPT = "score_ngrams_concat"
NUM_TOKENS = 100
```

```python
cmd = [
    str(PYTHON), "-u", str(SCRIPT),
    "--known_loc", KNOWN_LOC,
    "--unknown_loc", UNKNOWN_LOC,
    "--metadata_loc", METADATA_LOC,
    "--model_loc", MODEL_LOC,
    "--save_loc", SAVE_LOC,
    "--completed_loc", COMPLETED_LOC,
    "--corpus", CORPUS,
    "--data_type", DATA_TYPE,
    "--problem", PROBLEM,
    "--score_script", SCORE_SCRIPT,
    "--fixed_num_tokens", str(NUM_TOKENS),
    "--greatest_common"
]

print("Running:\n", " ".join(cmd))
```

```python
# run from repo root so relative paths inside the script behave as expected
proc = subprocess.run(
    cmd,
    cwd=str(REPO_ROOT),
    text=True,
    capture_output=True
)

print(proc.stdout)
if proc.returncode != 0:
    print(proc.stderr)
    raise RuntimeError(f"Script failed with exit code {proc.returncode}")
```
